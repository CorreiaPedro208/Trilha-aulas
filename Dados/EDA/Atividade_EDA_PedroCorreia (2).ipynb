### Observe valores muito diferentes

Um **outlier** é um valor muito distante dos demais. Ele pode ser um erro, mas também pode ser um caso raro e verdadeiro. O boxplot ajuda a localizar esses valores; ele não decide o que deve ser removido.


```python
# TODO: substitua pelo nome de uma coluna numérica.

coluna_boxplot = "median_house_value"

# monte um boxplot
plt.figure(figsize=(6, 7))
sns.boxplot(y=df[coluna_boxplot], width=0.4, showfliers=False)
sns.stripplot(y=df[coluna_boxplot].sample(300, random_state=1),
              color='crimson', alpha=0.4, size=3, jitter=0.25)
plt.title(f"Boxplot de {coluna_boxplot}")
plt.ylabel("Valor mediano da casa (US$)")
plt.tight_layout()
plt.show()

coluna_boxplot2 = "population"

plt.figure(figsize=(6, 7))
sns.boxplot(y=df[coluna_boxplot2], width=0.4,
            flierprops={'marker': 'o', 'markersize': 4,
                        'markerfacecolor': 'none', 'alpha': 0.4})
plt.title(f"Boxplot de {coluna_boxplot2} (20.640 distritos)")
plt.ylabel("População do distrito")
plt.tight_layout()
plt.show()

```

```python
colunas_num = ["median_house_value", "median_income", "housing_median_age", "population"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.flat, colunas_num):
    sns.histplot(df[col], bins=50, ax=ax)
    ax.set_title(f"Distribuição de {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Nº de distritos")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, y="ocean_proximity",
              order=df["ocean_proximity"].value_counts().index, ax=axes[0])
axes[0].set_title("Quantidade de distritos por proximidade do oceano")
axes[0].set_xlabel("Nº de distritos")
axes[0].set_ylabel("Proximidade do oceano")

ordem = df.groupby("ocean_proximity")["median_house_value"].median().sort_values().index
sns.barplot(data=df, y="ocean_proximity", x="median_house_value",
            order=ordem, estimator="median", errorbar=None, ax=axes[1])
axes[1].set_title("Valor mediano da casa por proximidade do oceano")
axes[1].set_xlabel("Valor mediano da casa (mil US$)")
axes[1].set_ylabel("Proximidade do oceano")



plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 8))
sns.scatterplot(data=df, x="longitude", y="latitude",
                hue="median_house_value", palette="viridis",
                alpha=0.4, s=12)
plt.title("Distribuição geográfica do valor das casas na Califórnia")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend(title="Valor da casa (US$)", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

from matplotlib.ticker import FuncFormatter

ordem = df.groupby("ocean_proximity")["median_house_value"].median().sort_values().index

plt.figure(figsize=(9, 5))
sns.barplot(data=df, y="ocean_proximity", x="median_house_value",
            order=ordem, estimator="median", errorbar=None)
plt.title("Valor mediano da casa por proximidade do oceano")
plt.xlabel("Valor mediano da casa (US$)")
plt.ylabel("Proximidade do oceano")
plt.gca().xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x/1000:.0f}k"))
plt.tight_layout()
plt.show()
```
